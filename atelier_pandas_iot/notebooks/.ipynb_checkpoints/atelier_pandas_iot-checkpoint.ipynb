**5) Créer, de deux façons différentes, une Series de 4 températures avec comme index les heures de mesure 12h, 13h, 14h et 15h**

```python
# 1ère façon : en passant l'index directement
temperatures_heures_1 = pd.Series([21.5, 22.8, 19.4, 23.1], index=["12h", "13h", "14h", "15h"])
temperatures_heures_1
```

```python
# 2e façon : en réaffectant l'index d'une Series existante
temperatures_heures_2 = temperatures.copy()
temperatures_heures_2.index = ["12h", "13h", "14h", "15h"]
temperatures_heures_2
```

**6) Afficher la 1ère et la dernière valeur**

```python
temperatures_heures_1.iloc[0], temperatures_heures_1.iloc[-1]
```

## Partie 5 – Manipulation des colonnes

**1) Ajouter la colonne "temperature_fahrenheit"**

```python
df["temperature_fahrenheit"] = df["temperature"] * 9 / 5 + 32
df.head()
```

**2) Ajouter la colonne "niveau_temperature" pour catégoriser la température en "Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy**

```python
df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
df[["temperature", "niveau_temperature"]].head()
```

**3) Renommer la colonne "humidite" en "humidite_relative"**

```python
df = df.rename(columns={"humidite": "humidite_relative"})
df.columns
```

**4) Supprimer la colonne "temperature_fahrenheit"**

```python
df = df.drop(columns=["temperature_fahrenheit"])
df.columns
```

## Partie 6 – Filtrage

**1) Récupérer uniquement les lignes telles que température > 30**

```python
df[df["temperature"] > 30]
```

**2) Récupérer uniquement les lignes telles température > 30 et humidité > 70**

```python
df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
```

## Partie 7 – Tri

**1) Créer df_trie1 pour trier en fonction de la température croissante**

```python
df_trie1 = df.sort_values("temperature")
df_trie1.head()
```

**2) Créer df_trie2 pour trier en fonction de la température décroissante**

```python
df_trie2 = df.sort_values("temperature", ascending=False)
df_trie2.head()
```

**3) Créer top10 pour voir les 10 températures les plus élevées**

```python
top10 = df.sort_values("temperature", ascending=False).head(10)
top10
```

**4) Trier en fonction des colonnes "batiment" (croissant) et "temperature" (décroissant)**

```python
df.sort_values(["batiment", "temperature"], ascending=[True, False])
```

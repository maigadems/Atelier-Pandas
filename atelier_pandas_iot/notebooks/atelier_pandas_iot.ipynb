## Partie 5 – Manipulation des colonnes

**1) Ajouter la colonne "temperature_fahrenheit"**

```python
df["temperature_fahrenheit"] = df["temperature"] * 9 / 5 + 32
df.head()
```

**2) Ajouter la colonne "niveau_temperature" pour catégoriser la température en "Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy**

```python
df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
df[["temperature", "niveau_temperature"]].head()
```

**3) Renommer la colonne "humidite" en "humidite_relative"**

```python
df = df.rename(columns={"humidite": "humidite_relative"})
df.columns
```

**4) Supprimer la colonne "temperature_fahrenheit"**

```python
df = df.drop(columns=["temperature_fahrenheit"])
df.columns
```

## Partie 6 – Filtrage

**1) Récupérer uniquement les lignes telles que température > 30**

```python
df[df["temperature"] > 30]
```

**2) Récupérer uniquement les lignes telles température > 30 et humidité > 70**

```python
df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
```

## Partie 9 – Gestion des valeurs manquantes

**1) Afficher le nombre de valeurs manquantes par colonne**

```python
df.isna().sum()
```

**2) Afficher le nombre total des valeurs manquantes**

```python
df.isna().sum().sum()
```

**3) Afficher le taux de valeurs manquantes**

```python
df.isna().mean() * 100
```

**4) Afficher les lignes contenant des valeurs manquantes**

```python
df[df.isna().any(axis=1)]
```

**5) Pour la température et l'humidité, remplacer les valeurs manquantes par les moyennes**

```python
df["temperature"] = df["temperature"].fillna(df["temperature"].mean())
df["humidite_relative"] = df["humidite_relative"].fillna(df["humidite_relative"].mean())
df[["temperature", "humidite_relative"]].isna().sum()
```

**6) Pour la consommation, remplacer les valeurs manquantes par la médiane**

```python
df["consommation"] = df["consommation"].fillna(df["consommation"].median())
df["consommation"].isna().sum()
```

**7) Pour "etat", remplacer les valeurs manquantes par "INCONNU"**

```python
df["etat"] = df["etat"].fillna("INCONNU")
df["etat"].isna().sum()
```
